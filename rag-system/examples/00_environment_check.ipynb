### 3. 환경 설정 확인

아래 셀들을 순서대로 실행하여 모든 환경이 올바르게 설정되었는지 확인하세요.

## 1. 필수 패키지 Import 확인

````python
# 필수 패키지 import 테스트
import os
import sys
from dotenv import load_dotenv
from langchain.chat_models import init_chat_model

print("모든 필수 패키지가 정상적으로 import 되었습니다.")
print(f"Python 버전: {sys.version}")
````

## 2. 환경 변수 로드 및 API 키 확인

````python
# .env 파일 로드
load_dotenv()

# API 키 확인
api_key = os.getenv("OPENAI_API_KEY")

if api_key:
    print("OPENAI_API_KEY가 정상적으로 로드되었습니다.")
    print("API Key ?? ???? ????.")
else:
    print("OPENAI_API_KEY를 찾을 수 없습니다.")
    print(".env 파일을 생성하고 OPENAI_API_KEY를 설정해주세요.")
````

## 3. OpenAI API 연결 테스트

````python
try:
    model = init_chat_model(
        "gpt-5.6-sol",
    )

    response = model.invoke("안녕 너는 누구야?")
    print(response.content)
except Exception as e:
    print(f"모델 초기화 또는 호출 중 오류 발생: {e}")
````

모든 셀이 정상적으로 실행되었다면 환경 설정이 완료된 것입니다!
